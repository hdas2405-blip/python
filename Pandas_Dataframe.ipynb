Q1. Creating a Pandas DataFrame using a Python dictionary with at least 5 rows of sample data

```python
import pandas as pd

# Creating the dictionary with required columns
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Printer'],
    'Region': ['North', 'South', 'East', 'West', 'North'],
    'Sales': [1200, 25, 45, 300, 150],
    'Quantity': [5, 50, 30, 10, 8]
}

# Converting to DataFrame
df_q1 = pd.DataFrame(data)

# Displaying the DataFrame
display(df_q1)
```

Dataset Setup for Questions 2-10

```python
# Loading the provided retail dataset for the remaining questions
df = pd.read_csv("Retail_dataset.csv")
```

Q2. Data Inspection

```python
# 1. Display the first 5 rows
print("First 5 rows:")
display(df.head())

# 2. Display the last 5 rows
print("\nLast 5 rows:")
display(df.tail())

# 3. Show the structure of the dataset using info()
print("\nDataset Structure:")
df.info()

# 4. Display the shape of the dataset
print("\nDataset Shape:", df.shape)
```

Q3. Column Selection

```python
# Selecting specific columns
selected_columns = df[['Product', 'Region', 'Sales']]
display(selected_columns.head()) # Using head() to keep output clean, remove if full output is desired
```

Q4. Indexing using .loc

```python
# 1. Display the row with index 10
print("Row with index 10:")
display(df.loc[10:10]) # Slicing used slightly to return a DataFrame instead of a Series for cleaner display

# 2. Display the Sales value of the row with index 20
sales_at_20 = df.loc[20, 'Sales']
print(f"\nSales value at index 20: {sales_at_20}")
```

Q5. Indexing using .iloc

```python
# 1. The first 6 rows and first 4 columns
print("First 6 rows and 4 columns:")
display(df.iloc[:6, :4])

# 2. The Sales value from the 3rd row (index 2)
# Assuming 'Sales' is dynamically located to ensure accuracy with iloc
sales_col_index = df.columns.get_loc('Sales')
sales_3rd_row = df.iloc[2, sales_col_index]
print(f"\nSales value from the 3rd row: {sales_3rd_row}")
```

Q6. Row Slicing

```python
# 1. The first 10 rows of the dataset
print("First 10 rows:")
display(df[:10])

# 2. Rows 15 to 25
print("\nRows 15 to 25:")
display(df[15:26]) # 26 is used to make the slice inclusive of index 25
```

Q7. Filtering Data

```python
# 1. Rows where Region = "East"
print("Region = 'East':")
display(df[df['Region'] == 'East'])

# 2. Rows where Sales > 1500
print("\nSales > 1500:")
display(df[df['Sales'] > 1500])
```

Q8. Multiple Condition Filtering

```python
# Multiple condition filtering using &
filtered_data = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]
display(filtered_data)
```

Q9. Descriptive Statistics

```python
# Generate summary statistics
stats = df.describe()
display(stats)

# Extracting specific values from the description
mean_sales = stats.loc['mean', 'Sales']
max_quantity = stats.loc['max', 'Quantity']
min_sales = stats.loc['min', 'Sales']

print(f"\nMean sales value: {mean_sales}")
print(f"Maximum quantity sold: {max_quantity}")
print(f"Minimum sales value: {min_sales}")
```

Q10. Aggregation Analysis

```python
# Aggregation on the Sales column
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
max_sales = df['Sales'].max()
min_sales = df['Sales'].min()

print(f"1. Total sales: {total_sales}")
print(f"2. Average sales: {average_sales}")
print(f"3. Maximum sales: {max_sales}")
print(f"4. Minimum sales: {min_sales}")
```
