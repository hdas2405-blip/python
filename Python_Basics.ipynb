Question 5 : Write a Python program to check whether a number is even or odd using conditional
statements.

```python
# Taking integer input from the user
number = int(input("Enter a number to check: "))

# Conditional statement to check for even or odd
if number % 2 == 0:
    print(f"{number} is an Even number.")
else:
    print(f"{number} is an Odd number.")
```

Question 6: Write a program to print numbers from 1 to 10 using a loop.

```python
# Using a for loop with the range function (stops at 11 to include 10)
for i in range(1, 11):
    print(i)
```

Question 7: Create a list of five numbers and print the maximum number from the list.

```python
# Creating a list of five numbers
numbers_list = [14, 52, 9, 87, 33]

# Finding and printing the maximum number using the built-in max() function
max_number = max(numbers_list)
print(f"The maximum number in the list is: {max_number}")
```

Question 8: Write a Python program to remove duplicate values from a list using a set.

```python
# Original list with duplicate values
list_with_duplicates = [1, 2, 2, 3, 4, 4, 5]

# Converting to a set automatically removes duplicates, then we convert back to a list
unique_list = list(set(list_with_duplicates))

print(f"Original List: {list_with_duplicates}")
print(f"List after removing duplicates: {unique_list}")
```

Question 9: Write a function that returns the square of a number.

```python
# Defining the function
def square_number(n):
    return n ** 2

# Testing the function
test_num = 6
result = square_number(test_num)
print(f"The square of {test_num} is {result}")
```

Question 10: Write a Python program to count how many times a number appears in a list. Example List [2,3,4,2,5,2]

```python
# Example list provided in the prompt
example_list = [2, 3, 4, 2, 5, 2]

# The number we want to count
target_number = 2

# Using the built-in count() method
appearance_count = example_list.count(target_number)

print(f"The number {target_number} appears {appearance_count} times in the list.")
```
