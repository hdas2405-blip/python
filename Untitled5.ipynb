```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (Make sure the CSV file is uploaded to your Colab session)
df = pd.read_csv('House Price India.csv')
df.head()
```

```python
# Check basic info (columns, data types, non-null counts)
print("--- Dataset Info ---")
df.info()

# Statistical summary of numerical columns
print("\n--- Dataset Description ---")
df.describe()
```

```python
# Check for missing values
print("--- Missing Values ---")
print(df.isnull().sum())

# If missing values exist, you can drop them or fill them.
# Example: Dropping missing values
df.dropna(inplace=True)

# Alternatively, filling missing numerical values with the median:
# df.fillna(df.median(numeric_only=True), inplace=True)
```

Q1. What is the average, median, and standard deviation of house prices?

```python
price_mean = df['Price'].mean() # Ensure 'Price' matches your column name
price_median = df['Price'].median()
price_std = df['Price'].std()

print(f"Average Price: {price_mean:.2f}")
print(f"Median Price: {price_median:.2f}")
print(f"Standard Deviation: {price_std:.2f}")
```

Interpretation: The average (mean) gives the general price level, while the median shows the middle value. If the mean is significantly higher than the median, it indicates that a few extremely expensive houses are pulling the average up. The standard deviation tells you how much the house prices vary from the average.

Q2. Which number of bedrooms is most common?

```python
most_common_beds = df['number of bedrooms'].mode()[0]
print(f"Most common number of bedrooms: {most_common_beds}")

# To see the full distribution:
print(df['number of bedrooms'].value_counts())
```

Interpretation: This identifies the primary market demand or the standard architectural design in the region. For example, if 3 is the most common, the market is primarily composed of standard family homes.

Q3. Check if the price data is skewed.

```python
price_skewness = df['Price'].skew()
print(f"Price Skewness: {price_skewness:.2f}")

# Visualizing the skewness
plt.figure(figsize=(8,5))
sns.histplot(df['Price'], kde=True)
plt.title('Distribution of House Prices')
plt.show()
```

Interpretation: A skewness value greater than 0 means the data is right-skewed (a long tail on the right), which is very common in real estate since a few luxury properties can cost significantly more than average homes. A value near 0 indicates a normal distribution.

Q4. What is the average price for each number of bedrooms?

```python
avg_price_per_bed = df.groupby('number of bedrooms')['Price'].mean().reset_index()
print(avg_price_per_bed)

# Visualizing the trend
plt.figure(figsize=(8,5))
sns.barplot(x='number of bedrooms', y='Price', data=avg_price_per_bed)
plt.title('Average Price vs. Number of Bedrooms')
plt.show()
```

Interpretation: This shows how adding a bedroom affects property value. Generally, price increases with the number of bedrooms, but you might notice a drop-off at very high numbers (e.g., 8+ bedrooms) if those properties are older, boarding houses, or in less desirable locations.

Q5. What is the relationship between living area and price? (Use: Area without Basement).

```python
# Q5. What is the relationship between living area and price?

# Using the exact column name from the dataset to prevent KeyErrors
area_col = 'Area of the house(excluding basement)'

# 1. Calculate the correlation numerically
correlation = df[area_col].corr(df['Price'])
print(f"Correlation between Area without Basement and Price: {correlation:.2f}")

# 2. Visualize the relationship using a scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(x=area_col, y='Price', data=df)

# Formatting the chart
plt.title('Q5: Living Area (without Basement) vs. Price')
plt.xlabel('Area without Basement (sq ft)')
plt.ylabel('Price')
plt.show()
```

Interpretation: A positive correlation closer to 1 indicates a strong direct relationship (as area increases, price increases). The scatter plot will visually confirm if the relationship is linear and highlight how tightly grouped the data points are.

Q6. Identify any anomalies where houses have high prices but low areas.

```python
# Define thresholds for "high price" (top 10%) and "low area" (bottom 10%)
area_col = 'Area of the house(excluding basement)'
high_price_threshold = df['Price'].quantile(0.90)
low_area_threshold = df[area_col].quantile(0.10)

# Filter the dataset for these specific conditions
anomalies = df[(df['Price'] > high_price_threshold) & (df[area_col] < low_area_threshold)]

print(f"Number of anomalies found: {len(anomalies)}")
display(anomalies.head())
```

Interpretation:
Finding these anomalies is crucial for data cleaning and business context. These data points represent outliers. In the real estate market, a house with very little square footage but a massive price tag usually means one of three things: it is an ultra-luxury micro-apartment in a highly sought-after downtown location, it is a property with significant historical value, or it is simply a data entry error that might need to be removed before training a machine learning model.

Q7. Compare average price based on number of floors and houses with or without waterfront.

```python
# Group by both floors and waterfront, then calculate the mean price
floor_water_comparison = df.groupby(['number of floors', 'waterfront present'])['Price'].mean().unstack()
print(floor_water_comparison)

# Visualize the comparison
floor_water_comparison.plot(kind='bar', figsize=(10, 6))
plt.title('Average Price by Floors and Waterfront Presence')
plt.xlabel('Number of Floors')
plt.ylabel('Average Price')
plt.legend(title='Waterfront Present (0=No, 1=Yes)')
plt.show()
```

Interpretation:
This multi-variable analysis reveals premium features in the housing market. The data will consistently show that homes with a waterfront command a massive price premium compared to non-waterfront homes, regardless of how many floors the house has. This indicates that geographic and scenic advantages often outweigh structural sizes in high-end real estate valuation.

Q8. Identify the minimum and maximum house price. What does this indicate?

```python
min_price = df['Price'].min()
max_price = df['Price'].max()
price_range = max_price - min_price

print(f"Minimum House Price: {min_price}")
print(f"Maximum House Price: {max_price}")
print(f"Total Price Range: {price_range}")
```

Interpretation:
The massive gap between the minimum and maximum prices highlights the extreme wealth disparity and diversity of the real estate market captured in this dataset. It indicates that the dataset is not localized to a single type of neighborhood, but rather encompasses everything from lower-income or highly distressed properties to ultra-luxury estates.

Q9. Which location (zipcode/area) has the highest average price?

```python
# Group by Postal Code, calculate average price, and sort descending
highest_avg_price_location = df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False)

# Extract the top location and its price
top_zipcode = highest_avg_price_location.index[0]
top_price = highest_avg_price_location.iloc[0]

print(f"Location (Postal Code) with the highest average price: {top_zipcode}")
print(f"Average Price in this location: {top_price:.2f}")
```

Interpretation:
This isolates the most exclusive, premium neighborhood in the dataset. Real estate is heavily dependent on location. Identifying the highest-priced zip code shows where market demand is strongest, which is likely driven by factors not explicitly in the dataset, such as school districts, low crime rates, or proximity to commercial hubs.
