```python
import pandas as pd
df = pd.read_csv("/content/Advanced_pandas_sales_dataset.csv")
```

Q1 Dataset Exploration

```python
# Display the first 10 rows [cite: 19]
print("First 10 rows:")
display(df.head(10))

# Display the dataset structure using info() [cite: 20]
print("\nDataset Structure:")
df.info()

# Find the number of rows and columns [cite: 20]
print(f"\nNumber of Rows: {df.shape[0]}, Number of Columns: {df.shape[1]}")

# Identify the data types of each column [cite: 20]
print("\nData Types:")
print(df.dtypes)
```

Q2 Sales Filtering Analysis

```python
# 1. Orders where Sales > 3500 [cite: 22]
high_sales = df[df['Sales'] > 3500]
print("Orders with Sales > 3500:")
display(high_sales)

# 2. Orders where Region = "East" and Sales > 2500
east_high_sales = df[(df['Region'] == 'East') & (df['Sales'] > 2500)]
print("\nEast Region Orders with Sales > 2500:")
display(east_high_sales)
```

Q3 Customer Purchase Analysis

```python
# Group by Customer, sum the Sales, and get the top 5
top_customers = df.groupby('Customer')['Sales'].sum().nlargest(5).reset_index()
print("Top 5 Customers by Total Sales:")
display(top_customers)
```

Q4 Regional Sales Performance

```python
# Calculate the total sales generated by each region [cite: 28]
regional_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False).reset_index()
print("Total Sales by Region:")
display(regional_sales)

# Which region generated the highest revenue? [cite: 29]
highest_region = regional_sales.iloc[0]['Region']
print(f"\nThe region that generated the highest revenue is: {highest_region}")
```

Q5 Product Performance Analysis

```python
# Calculate the total quantity sold for each product
product_qty = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False).reset_index()
print("Total Quantity Sold per Product:")
display(product_qty)

# Which product appears to be the most popular?
most_popular_product = product_qty.iloc[0]['Product']
print(f"\nThe most popular product is: {most_popular_product}")
```

Q6 Multi-Level Grouping

```python
# Group the dataset by: Region and Category [cite: 33]
# Calculate the total sales for each combination
region_category_sales = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()
print("Total Sales by Region and Category:")
display(region_category_sales)
```

Q7 Pivot Table Creation

```python
# Create a pivot table showing Region (Rows), Product (Columns), and Total Sales (Values) [cite: 40, 41, 42, 43]
pivot_sales = pd.pivot_table(df, values='Sales', index='Region', columns='Product', aggfunc='sum', fill_value=0)
print("Pivot Table: Sales by Region and Product")
display(pivot_sales)

# Which region sells the most laptops? [cite: 44]
# Assuming the product column contains a value like 'Laptop'
if 'Laptop' in pivot_sales.columns:
    laptop_top_region = pivot_sales['Laptop'].idxmax()
    print(f"\nThe region that sells the most laptops is: {laptop_top_region}")
```

Q8 Profitability Analysis

```python
# Calculate the average profit for each product category
avg_profit_category = df.groupby('Category')['Profit'].mean().sort_values(ascending=False).reset_index()
print("Average Profit by Category:")
display(avg_profit_category)

# Which category is most profitable?
most_profitable_category = avg_profit_category.iloc[0]['Category']
print(f"\nThe most profitable category is: {most_profitable_category}")
```

Q9 Discount Impact Study

```python
# Calculate the average profit for each discount level
discount_profit = df.groupby('Discount')['Profit'].mean().sort_values(ascending=False).reset_index()
print("Average Profit by Discount Level:")
display(discount_profit)
```

Q10 End-to-End Data Wrangling Task

```python
# 1. Filter the dataset where Sales > 3000 [cite: 51]
filtered_df = df[df['Sales'] > 3000]

# 2. Group data by Region and Product [cite: 52]
# 3. Calculate total sales [cite: 52]
# 4. Reshape the result into a pivot table [cite: 53]
end_to_end_pivot = pd.pivot_table(filtered_df, values='Sales', index='Region', columns='Product', aggfunc='sum', fill_value=0)

print("End-to-End Data Wrangling Pivot Table:")
display(end_to_end_pivot)
```
